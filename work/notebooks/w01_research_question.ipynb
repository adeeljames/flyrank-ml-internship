```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

declining_rate = df["trend_direction"].str.lower().eq("down").mean()
stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).mean()
median_impressions = df["impressions_90d"].median()

print(f"Declining rate (trend_direction == 'down'): {declining_rate:.3f}")
print(f"Share of pages stale AND visible (>=180 days, >=500 impressions): {stale_visible:.3f}")
print(f"Median impressions_90d across all pages: {median_impressions:.0f}")
```

Observed: 54.2% of pages in this sample are currently in a "down" trend — a substantial
portion of the inventory, so a prioritization tool has real, ongoing work to do.

Only 0.1% of pages are both stale (180+ days since update) AND highly visible (500+
impressions/90d) — this combination is rare, which tells me the "stale AND visible"
rule alone is too narrow to be the primary signal; the lane will need broader or
combined signals (e.g. lower impression thresholds, or combining staleness with
declining trend) rather than relying on this single rule.

The median page gets 731 impressions over 90 days, showing there's meaningful search
visibility across the inventory worth protecting or improving.

This is direct evidence that: (a) there's a large enough declining population to
justify a ranking approach, and (b) simple single-condition rules like "stale AND
visible" are too strict on their own — reinforcing why a learned model (as shown in
the starter pipeline: Precision@50 = 0.740 vs baseline 0.240) adds real value over a
single hand-written threshold.
